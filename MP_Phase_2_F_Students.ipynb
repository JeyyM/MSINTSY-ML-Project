**Description**

Dataset that contains the population of the major denominations of the most popular religions such as Christianity, Judaism, Islam, Buddhism, Hinduism, Atheists, and other religions such as Folk Religions like Animism for every country across the years from 1945 to 2010.

**Collection Method**

The dataset was collected by the World Religion Project (WRP) who collected data on global religious adherence from 1945 to 2010 in five-year intervals. It provides both raw counts and population percentages of adherents to various religions across many countries internationally. The data was compiled in three main stages:

Classification Development: A comprehensive literature review was conducted to define religion, identify measurable indicators, and structure a hierarchical religion tree categorizing major religions and their denominations.

Data Collection: Researchers gathered data from major existing sources, organizing it according to the religion tree. This resulted in multiple records per state per year, along with numerous gaps.

Data Cleaning and Imputation: The dataset was refined by reconciling discrepancies across sources and filling in missing data through imputation methods.

**Implications**
1. Estimated and Imputed Data: The dataset was compiled from multiple sources, then cleaned and organized, with missing values imputed by researchers. This imputation process may introduce inaccuracies, as it relies on assumptions and subjective decisions that can reflect the researchers' own biases. As a result, some figures may be distorted or artificially smoothed, potentially differing from the actual historical trends.

2. Variation is Reliability of Sources: Data was pulled from multiple sources, each with its own methodology and level of accuracy. As a result, the rigor and standards of data collection may vary across countries and regions. This could lead to differences in data quality, with some countries having more complete and accurate records than others.

3. Five-Year Intervals: The data may not be able to capture short-term shifts and events. This may lead to missing or misrepresentations of major historical events or migrations.

# **C. GDP  1960 to 2023.csv**

**Description**

Dataset that contains the GDP per Capita (in US Dollars) of most countries from 1960 to 2023 as collected by the World Bank Group. Choosing GDP per capita allows for a more accurate representation of wealth since it balances the economic output of its country with its total population. This more accurately shows the prosperity of each citizen.

**Collection Method**

The data is compiled by the World Bank Group, which sources its data from the World Bank national accounts data and OECD National Accounts data files. GDP is calculated as the sum of gross value added by all resident producers in the economy, plus any product taxes, and minus any subsidies not included in the value of the products. The calculation does not account for the depreciation of fabricated assets, such as equipment and machinery, or for the depletion and degradation of natural resources.

After calculating GDP values in local currencies, the figures are converted to U.S. dollars using the official exchange rate for that specific year. In cases where the official exchange rate does not accurately reflect market conditions, an alternative conversion factor is applied, although the exact method for choosing this alternative is not specified.

It uses the gap-filled total method for aggregation and imputation. In this approach, missing values are estimated based on the relationship between a country's data a benchmark year, usually 2010. These estimates are used to fill in data both forward and backward from that year. If data for 2010 is missing, the World Bank uses proxy variables such as gross national income, population, exports and imports, and value added in agriculture, industry, and services to estimate the missing values. However, imputation is generally not performed unless missing data accounts for more than one-third of the total in the benchmark year.


**Implications**
1. Estimated and Imputed Data: For countries that lack data, values could be distorted compared to reality. This will most likely affect lower income or countries in crises. This may result in developed countries being more stable or less developed countries to be more volatile.

2. GDP as a Full Proxy for Wellbeing: Though GDP is a valuable source for the development of a country, it only accounts for the economics of goods and services. It does not take into account the quality of life of its population or if its GDP is the result of unscrupulus practices.

3. Unclear Wealth Distribution: GDP does not reflect how the wealth is shared across the population. Despite a high GDP, a country's wealth may be concentrated among its elites, leading to a lower quality of life overall for the majority.

# **D. Human Development Index 1990 to 2020.csv**

**Description**

The Human Development Index (HDI) measures a country's average achievements in health, education, and standard of living using a composite of normalized indicators. It uses life expectancy, years of schooling, and gross national income (GNI) per capita along with measures in inequality for education, income, and gender.

**Collection Method**

The Human Development Reports collect data from various agencies for different factors such as:

* Health-Related: Reports from the United Nations Department of Economic and Social Affairs (UN DESA)
* Education-Related: Reports from Barro & Lee, United Nations Educational, Scientific and Cultural Organization (UN ESCO), Demographic and Health Surveys (DHS), UNICEF, and more.
* Economics-Related: Reports from International Monetary Fund (IMF), United Nations Statistics Division, UN DESA, and World Bank

The values for features are limited from a range of 0 to 1 as goalposts where one is the highest possible. The overall Human Development Index is formed from the normalized values of various features where certain factors such as wealth like GNI are logarithmically adjusted to make all factors fair. These different features form the indices in the report such as the Human Development Index (HDI), Gender Development Index (GDI), Multidimensional Poverty Index (MPI), and more.

**Implications**
1. New Factor Data Gaps: The dataset introduces various factors such as "Inequality-adjusted Human Development Index" which was only introduced in 2010 which may limit potential factors to examine due to lack of data.

2. Data Gaps for Developing Countries: Developing countries may have gaps for features especially in the early years as well as for newer features.

3. Potential Changes in Measurement: Though not stated, methodologies for each feature's measurement may have changed over the past 30 years which may affect comparisons.

### Analysis of Dominant Religion Change

Since dominant religion will be a major factor in the analysis of each religion's performance, it is important to analyze if religious change is a common occurence.

**Process**
1. Check each country's total population for each year to remove duplicate entries for every religion.
2. Use the mean of their total population to maintain the same population values.
3. Filter to only get years < 2025 to only consider up to the present day.

4. Once simplified, count the number of unique dominant religions.
    - Make sure blank Dominants are not counted as a change
5. Find only those who did change dominant religion
6. Just consider five-year intervals since values inbetween are imputed

7. Find the number of unique religions overall to find patterns

```python
4# To get a simplified version to easily analyze if changes occured
dominance_change_summary = (
    df_individual_countries_population.groupby(['Country', 'Year'])
    .agg({
        'Value_interp': 'sum',
        'Dominant': 'first'
    }).reset_index().rename(columns={'Value_interp': 'Total_Population'})
)

# display(dominance_change_summary)

dominance_change_summary_filtered = dominance_change_summary[dominance_change_summary['Year'] < 2025]

dominant_counts = (
    dominance_change_summary_filtered.dropna(subset=['Dominant']).groupby('Country')['Dominant'].nunique()
)

countries_with_change = dominant_counts[dominant_counts > 1].index.tolist()

print("\n--- Countries whose dominant religion changed ---")
countries_with_change_df = pd.DataFrame(countries_with_change, columns=['Country'])
display(countries_with_change_df)

summary_changed = dominance_change_summary[
    dominance_change_summary['Country'].isin(countries_with_change)
]

summary_changed_every5 = (
    summary_changed[summary_changed['Year'] % 5 == 0]
    .sort_values(['Country', 'Year'])
)

print("\n--- Timeline every 5 years ---")
display(summary_changed_every5)

dominant_counts = (
    summary_changed_every5.groupby('Country')['Dominant'].nunique().reset_index()
)

dominant_counts_changed = dominant_counts[
    dominant_counts['Dominant'] > 1
]

print("\n--- Unique Religion Count ---")
display(dominant_counts_changed)
```

**Analysis**

In total, only 13 out of 198 countries experienced a change in their dominant religion between 1945 and 2025. This suggests that for the vast majority of countries, the dominant religion has remained stable over time.


The next important step is to examine the net gains and losses in dominance for each major religion. By quantifying which religions lost or gained influence globally, we can identify larger trends or whether certain religions are significantly expanding their dominant status across nations. This will help determine whether there are enough meaningful shifts in dominant religion to justify further analysis of whether religious change correlates with development.

**Process**
This is done by looping through each country and its row then making a record of changes such as the years of change along with from and to.

```python
changed_df = dominance_change_summary_filtered.copy()

countries_to_check = dominant_counts_changed['Country'].tolist()
changed_df = changed_df[changed_df['Country'].isin(countries_to_check)].dropna(subset=['Dominant'])

change_entries = []

for country, group in changed_df.groupby('Country'):
    timeline = group.sort_values('Year')
    prev_religion = None

    for index, row in timeline.iterrows():
        current_religion = row['Dominant']
        year = row['Year']

        if prev_religion is None:
            prev_religion = current_religion
            prev_year = year
            continue

        if current_religion != prev_religion:
            change_entries.append({
                'Country': country,
                'Change Year': year,
                'From': prev_religion,
                'To': current_religion
            })
            prev_religion = current_religion
            prev_year = year

changes_df = pd.DataFrame(change_entries)
display(changes_df)
```

**Analysis of Dominance Change**

Most of these countries changed only once, indicating a relatively stable long-term religious dominance. However, a few countries showed evidence of back-and-forth changes between two dominant religions. South Korea stands out as the only country in the dataset with three unique dominant religions over the timeline.

The next step is to view the changes in the count of dominant religions. This is done by adding and subtracting a counter for each To and From

```python
religion_tally = {}

for index, row in changes_df.iterrows():
    from_religion = row['From']
    to_religion = row['To']

    religion_tally[from_religion] = religion_tally.get(from_religion, 0) - 1

    religion_tally[to_religion] = religion_tally.get(to_religion, 0) + 1

religion_tally_df = pd.DataFrame(list(religion_tally.items()), columns=['Religion', 'Net_Change'])
religion_tally_df = religion_tally_df.sort_values('Net_Change', ascending=False).reset_index(drop=True)

print(religion_tally_df)
```

**Analysis**

The final net change tally shows that the overall dominance of religions across countries remains very stable. While individual countries did experience shifts in their dominant religion, the total net changes reveal that only one country shifted away from Islam to Unaffiliated, and the other major religions (Christianity, Buddhism) saw no significant net gains or losses in their number of country positions.

This result suggests that, on a globally, dominant religions tend to be set. This relative stability means that there may be too few instances of significant religious dominance change to meaningfully test whether such changes correlate with patterns in development.

# **Created Tier Values**

To create clear categories of overall success, countries were grouped into quintiles at five-year intervals, with each quintile representing 20% of the distribution. Each country was assigned a tier value from 1 to 5, where 5 indicates the highest level of success. These tier values were then tracked over time to measure mobility, capturing whether a country moved up or down in rank in subsequent periods.

This approach enables a deeper understanding of economic stratification, allowing the analysis to explore patterns of upward or downward mobility within and across tiers. It also facilitates the examination of how dominant religions are distributed across success tiers, helping to assess the proportion of successful countries associated with each religious group.

```python
# Drop NaN GDPs
newGDP_df = newGDP_df.dropna(subset=['GDP_per_capita']).reset_index(drop=True)
newGDP_df = newGDP_df.sort_values(by=['Country', 'Year']).reset_index(drop=True)

display(newGDP_df)

# Only keep years that match your tiering (every 5 years)
cleanedGDP = newGDP_df[newGDP_df['Year'] % 5 == 0].reset_index(drop=True)

# Setting tier values
# Tier values are just set for 5 year intervals, so 1961 or something won't have any just to make it easy
# Just filter by year % 5 == 0 and also those with 0 are blanks
year_range = np.arange(1960, 2021, 5)

processed = []

# Tier values are just set for 5-year intervals, so 1961 or something doesn't need tier values
for year in year_range:
    curr_group = cleanedGDP[cleanedGDP['Year'] == year].copy()

    gdp = curr_group['GDP_per_capita']

    if gdp.nunique() < 5:
        curr_group['Tier_Value'] = 0
    else:
        try:
            curr_group['Tier_Value'] = pd.qcut(
                gdp.rank(method='first'),
                q=5,
                labels=[1, 2, 3, 4, 5]
            ).astype(int)
        except Exception as e:
            print(f"Year {year} failed: {e}")
            curr_group['Tier_Value'] = 0

    processed.append(curr_group)

tiered_countries = pd.concat(processed, ignore_index=True)

# How much each country moved up/down compared to previous interval.
mobility = []
prev_tiers = {}

for year in year_range:
    curr_group = tiered_countries[tiered_countries['Year'] == year].copy()
    curr_group = curr_group[curr_group['Tier_Value'] != 0]

    tiers = curr_group['Tier_Value']
    mobility_values = []

    for index, row in curr_group.iterrows():
        country = row['Country']
        current_tier = row['Tier_Value']
        previous_tier = prev_tiers.get(country)

        if previous_tier is None:
            mobility_values.append(0)  # No previous, so no change
        else:
            mobility_values.append(current_tier - previous_tier)

        prev_tiers[country] = current_tier

    curr_group['Mobility_Value'] = mobility_values

    mobility.append(curr_group)

mobility_countries = pd.concat(mobility, ignore_index=True)

# Merge mobility back to tiered data
country_progression = tiered_countries.merge(
    mobility_countries[['Country', 'Year', 'Mobility_Value']],
    on=['Country', 'Year'],
    how='left'
)

# Ensure the column exists even if there are gaps
if 'Mobility_Value' not in country_progression.columns:
    country_progression['Mobility_Value'] = 0

# Merge back to original
newGDP_df = newGDP_df.merge(
    country_progression[['Country', 'Year', 'Tier_Value', 'Mobility_Value']],
    on=['Country', 'Year'],
    how='left'
)

# Fill in blanks
newGDP_df['Tier_Value'] = newGDP_df['Tier_Value'].fillna(0).astype(int)
newGDP_df['Mobility_Value'] = newGDP_df['Mobility_Value'].fillna(0)

newGDP_df = newGDP_df.sort_values(by=['Country', 'Year']).reset_index(drop=True)

mobility_df = newGDP_df.copy()
mobility_df = mobility_df.sort_values(by=['Country', 'Year']).reset_index(drop=True)

# print(newGDP_df.head(10))
display(mobility_df[50:100])
```

# Group Mean Comparison for GDP per Capita and Dominant Religion
**Process**

To be able to formally conclude that the dominant religions do perform significantly differently from each other in terms of economic outcomes, it was necessary to test whether the observed differences in GDP per capita across religious groups were statistically significant rather than due to random variation.

Using the merged dataset containing data on the yearly GDP per capita and dominant religion of a country, the group explored whether GDP per capita significantly varied by religion over time. The data was divided into 5 year intervals from 1960 to 2014. For each interval, the dataset was filtered to keep only entries with complete values for GDP and dominant religion.

Countries were then grouped based on their dominant religion, and two assumption checks were performed: the Shapiro-Wilk test for normality on each group and Levene's test for equal variances for across the groups. If all religion groups passed both assumptions, a one-way ANOVA test was applied to compare mean GDP per capita. If either assumption failed, the non-parametric Kruskal-Wallis test was used instead to compare medians. The test used, along with its F- or H-statistic and corresponding p-value, was recorded for each time interval. This approach allowed the group to account for the distributional characteristics of the data before selecting the appropriate test. A line graph was later produced to visualize the H-statistics across time intervals, with visual indicators highlighting intervals where statistically significant differences were seen.

```python
results = []
merged_df = combined_gdp_pop_df

for start in range(1960, 2015, 5):
    end = start + 4
    interval_label = f"{start}-{end}"

    print(f"\nInterval: {interval_label}")

    subset = merged_df[(merged_df['Year'] >= start) & (merged_df['Year'] <= end)]
    subset = subset.dropna(subset=['GDP_per_capita', 'Dominant'])

    if subset.empty:
        print("No data available.")
        results.append({'Interval': interval_label, 'Test': None, 'F-statistic': None, 'p-value': None})
        continue

    #Group by Dominant religion
    grouped = subset.groupby('Dominant')['GDP_per_capita']
    gdp_lists = [group for _, group in grouped if len(group) > 1]
    group_labels = [name for name, group in grouped if len(group) > 1]

    if len(gdp_lists) < 2:
        print("Not enough groups with valid data.")
        results.append({'Interval': interval_label, 'Test': 'N/A', 'F-statistic': None, 'p-value': None})
        continue

    #Shapiro-Wilk Test
    shapiro_results = [shapiro(group) for group in gdp_lists]
    normality_pvals = [round(res.pvalue, 4) for res in shapiro_results]

    print("Shapiro-Wilk p-values:")
    for name, p in zip(group_labels, normality_pvals):
        print(f"  - {name}: p = {p}")

    normal = all(p > 0.05 for p in normality_pvals)

    #Levene’s Test
    levene_stat, levene_p = levene(*gdp_lists)
    print(f"Levene’s Test p-value: {round(levene_p, 4)}")

    equal_var = levene_p > 0.05

    #Choose statistical test
    if normal and equal_var:
        f_stat, p_val = f_oneway(*gdp_lists)
        test_used = "ANOVA"
    else:
        f_stat, p_val = kruskal(*gdp_lists)
        test_used = "Kruskal-Wallis"

    print(f"{test_used} result: H = {round(f_stat, 4)}, p = {round(p_val, 4)}")

    results.append({
        'Interval': interval_label,
        'Test': test_used,
        'H-statistic': f_stat,
        'p-value': p_val
    })

#show final results table
print('\n')
anova_results = pd.DataFrame(results)
display(anova_results)
```

```python
anova_results['Significant'] = anova_results['p-value'] < 0.05

plt.figure(figsize=(10, 6))
sns.lineplot(x='Interval', y='H-statistic', data=anova_results, color='gray', label='H-statistic trend')
sns.scatterplot(
    x='Interval',
    y='H-statistic',
    data=anova_results,
    hue='Significant',
    palette={True: 'red', False: 'blue'},
    legend='brief',
    s=100
)
plt.xticks(rotation=45)
plt.title('Kruskal-Wallis H-statistics Over Time\n(Colored by p-value < 0.05)')
plt.ylabel('H-statistic')
plt.xlabel('Time Interval')
plt.legend(title='Significant (p < 0.05)', loc='upper right')
plt.tight_layout()
plt.show()
```

# **Chi-Square Test of Independence**

A Chi-Square Test of Independence was carried out to determine if certain religions were more commonly associated with specific economic tiers. This analysis is important because it allows the study to go beyond individual case observations and test whether the observed distribution of economic success across religions is due to chance or if there is a statistically meaningful pattern. The test assesses whether countries with a dominant religion are disproportionately represented in certain economic strata.

This matters because it addresses one of the study's core questions: whether a relationship exists between a country's dominant religion and its level of economic development. If a significant association is found, it could suggest that religion may potentially influence or correlate with national prosperity. Conversely, if no significant relationship is found, it would support the idea that economic success is largely independent of religious majority.

**Process**

To get appropriate results from the Chi-Square test, the group interpolated missing values of the original dataframe 'chi_df' for each country based on the existing entries. The group made these interpolations because some countries do not have complete observations, meaning not all entries corresponding to 1960, 1965, 1970, …, and 2010 are present have a tier value. This could lead to an undercounting of less well-off countries. For instance, Afghanistan only has 2000, 2005, and 2010 and thus, the group interpolated to include 1995, 1990, …, to 1960. The way the group interpolated column values is as follows:
- Country: country names from existing entries
- Year: years that are multiples of 5 but not present from existing entries, ranging from 1960 to 2010
- Tier-Value: mode from existing entries
- Dominant: mode from existing entries
- Total Population: for each interpolation, get 95% of its preceding value (5 years ahead) then round down
- Dominant_Religion_Population: for each interpolation, get 95% of its preceding value (5 years ahead) then round down
- Dominant_Population_Sum: for each interpolation, get 95% of its preceding value (5 years ahead) then round down
- Homogeneity: mean from existing entries
The interpolated version of 'chi_df' is stored in another dataframe variable 'interpolated_chi_df'.


```python
# 1. Prepare the base: get unique Country-Year-Dominant-Total Pop-Dominant Share
country_dominant = (
    religion_full_df
    .query('Year >= 1960')
    [['Country', 'Year', 'Dominant', 'total_population', 'dominant_share']]
    .drop_duplicates()
    .dropna()
)

# Compute the actual dominant religion population
country_dominant['dominant_religion_population'] = country_dominant['total_population'] * country_dominant['dominant_share']

# 2. Prepare your tier dataset
chi_df = tiered_countries.query('Year >= 1960').drop(['GDP_per_capita'], axis=1)
chi_df = chi_df.merge(country_dominant, on=['Country', 'Year'], how="left").dropna()

display(chi_df.head())

# 3. Interpolate missing years backward for each country
interpolated_chi_df = pd.DataFrame()
country_list = chi_df['Country'].unique()

for country in country_list:
    sub_df = chi_df[chi_df['Country'] == country].sort_values(by='Year', ascending=False).reset_index(drop=True)

    # Mode and base values
    tier_value = sub_df['Tier_Value'].mode().iloc[0]
    dominant = sub_df['Dominant'].mode().iloc[0]
    total_pop = sub_df['total_population'].iloc[0]
    dominant_pop = sub_df['dominant_religion_population'].iloc[0]
    dominant_share = dominant_pop / total_pop if total_pop != 0 else 0
    homogeneity = sub_df['Homogeneity'].mean() if 'Homogeneity' in sub_df else None

    # Years present
    year_list = set(sub_df['Year'])

    for year in range(2010, 1959, -5):
        if year not in year_list:
            new_row = {
                'Country': country,
                'Year': year,
                'Tier_Value': tier_value,
                'Dominant': dominant,
                'total_population': total_pop,
                'dominant_religion_population': dominant_pop,
                'dominant_share': dominant_share,
                'Homogeneity': homogeneity
            }
            sub_df = pd.concat([sub_df, pd.DataFrame([new_row])], ignore_index=True)

            # Decay population backward by 5% each missing step
            total_pop *= 0.95
            dominant_pop *= 0.95

    sub_df = sub_df.sort_values(by='Year').reset_index(drop=True)
    interpolated_chi_df = pd.concat([interpolated_chi_df, sub_df], ignore_index=True)

# 4. Inspect results
print('[Countries per year]')
for year in range(1960, 2011, 5):
    print(f"{year}: {len(interpolated_chi_df[interpolated_chi_df['Year'] == year])}")

print('\n[Entries per dominant religion]')
for religion in interpolated_chi_df['Dominant'].unique():
    count = len(interpolated_chi_df[interpolated_chi_df['Dominant'] == religion])
    print(f"{religion}: {count}")

display(interpolated_chi_df.head())
```

After that, they calculated summary statistics to be stored in the ‘tier_gdp_summary_stats’ variable, then converted to a dataframe variable as ‘tier_gdp_summary_df’. The complete list below shows columns corresponding to a set of summary statistics for each dominant religion per tier:

For each tier to be used in data modelling, the group plotted Spearman correlations, a correlation bar chart, and a GDP distribution histogram.


```python
# Filter out no tier value/gdp per capita
gdp_tier_analysis = religion_gdp_combination[
    (religion_gdp_combination['Tier_Value'] != 0) &
    (~religion_gdp_combination['GDP_per_capita'].isna())
]

# Group by Country, Year, Dominant, and Tier_Value, this makes it so countries get equally represented later
grouped = (
    gdp_tier_analysis
    .groupby(['Country', 'Year', 'Dominant', 'Tier_Value'], as_index=False)
    .agg({'GDP_per_capita': 'mean'})
)

# Get the means of each country, dominant, and tier. Each country will get a new entry for different tiers
country_means = (
    grouped
    .groupby(['Country', 'Dominant', 'Tier_Value'], as_index=False)
    .agg({'GDP_per_capita': 'mean'})
    .rename(columns={'GDP_per_capita': 'Country_Mean_GDP'})
)

# Calculate summary statistics
tier_gdp_summary_stats = []

for tier_value in sorted(country_means['Tier_Value'].dropna().unique()):
    tier_group = country_means[country_means['Tier_Value'] == tier_value]

    for religion, group in tier_group.groupby('Dominant'):
        data = group['Country_Mean_GDP'].dropna()

        if len(data) == 0:
            continue

        mean = data.mean()
        trimmed_mean = stats.trim_mean(data, 0.15)
        median = data.median()
        std = data.std()
        var = data.var()
        minGdp = data.min()
        maxGdp = data.max()
        rng = maxGdp - minGdp
        skew = data.skew()

        Q1 = data.quantile(0.25)
        Q3 = data.quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        outliers = data[(data < lower_bound) | (data > upper_bound)]
        total_outliers = len(outliers)
        percent_outliers = total_outliers / len(data) * 100

        outliers_sum = outliers.sum()
        total_sum = data.sum()
        percent_of_total_mean_from_outliers = (outliers_sum / total_sum) * 100 if total_sum != 0 else 0

        tier_gdp_summary_stats.append({
            'Tier_Value': tier_value,
            'Dominant_Religion': religion,
            'Count': len(data),
            'Mean': mean,
            'Trimmed Mean': trimmed_mean,
            'Median': median,
            'StdDev': std,
            'Variance': var,
            'Min': minGdp,
            'Max': maxGdp,
            'Range': rng,
            'Skew': skew,
            'Q1': Q1,
            'Q3': Q3,
            'IQR': IQR,
            'Lower_Bound': lower_bound,
            'Upper_Bound': upper_bound,
            'total_outliers': total_outliers,
            'percent_outliers': percent_outliers,
            'Outlier_Sum': outliers_sum,
            'Total_Sum': total_sum,
            'Percent_of_Mean_From_Outliers': percent_of_total_mean_from_outliers
        })

tier_gdp_summary_stats_df = pd.DataFrame(tier_gdp_summary_stats)

# Tier-wise outputs
print("\nTier-wise Detailed Summary with Plots\n")

for tier in sorted(tier_gdp_summary_stats_df['Tier_Value'].unique()):
    subset_stats = tier_gdp_summary_stats_df[tier_gdp_summary_stats_df['Tier_Value'] == tier]
    subset_means = country_means[country_means['Tier_Value'] == tier]

    print(f"========== TIER {tier} ==========\n")

    print("Central Tendency\n")
    display(subset_stats[['Tier_Value', 'Dominant_Religion', 'Count', 'Mean', 'Median', 'Trimmed Mean']])
    print("\n" + "-"*30 + "\n")

    print("Measures of Spread\n")
    display(subset_stats[['Tier_Value', 'Dominant_Religion', 'StdDev', 'Variance', 'Min', 'Max', 'Range', 'Skew']])
    print("\n" + "-"*30 + "\n")

    print("IQR & Outlier Bounds\n")
    display(subset_stats[['Tier_Value', 'Dominant_Religion', 'Q1', 'Q3', 'IQR', 'Lower_Bound', 'Upper_Bound']])
    print("\n" + "-"*30 + "\n")

    print("Outliers and their Impact\n")
    display(subset_stats[
        [
            'Tier_Value',
            'Dominant_Religion',
            'Count',
            'total_outliers',
            'percent_outliers',
            'Outlier_Sum',
            'Total_Sum',
            'Percent_of_Mean_From_Outliers'
        ]
    ])
    print("\n" + "="*50 + "\n")

    # Spearman correlations within this tier
    tier_correlations = []

    for religion in subset_means['Dominant'].unique():
        sub_corr_df = gdp_tier_analysis[
            (gdp_tier_analysis['Tier_Value'] == tier) &
            (gdp_tier_analysis['Dominant'] == religion)
        ][['dominant_share', 'GDP_per_capita']].dropna()

        if len(sub_corr_df) >= 2:
            corr, pval = spearmanr(sub_corr_df['dominant_share'], sub_corr_df['GDP_per_capita'])
            tier_correlations.append({
                'Dominant_Religion': religion,
                'Correlation': corr,
                'p_value': pval
            })

    tier_corr_df = pd.DataFrame(tier_correlations).sort_values('Correlation', ascending=False)
    print(f"Spearman Correlation within Tier {tier}\n")
    display(tier_corr_df)

    # Plot the correlation bar chart
    if not tier_corr_df.empty:
        plt.figure(figsize=(8, 4))
        sns.barplot(
            data=tier_corr_df,
            x='Correlation',
            y='Dominant_Religion',
            orient='h',
            palette='coolwarm'
        )
        plt.title(f'Tier {tier} | Spearman Correlation: Dominant Share vs GDP per Capita')
        plt.xlabel('Spearman Correlation')
        plt.xlim(-1, 1)
        plt.axvline(0, color='black', linestyle='--')

        for index, row in tier_corr_df.iterrows():
            plt.text(row['Correlation'], index, f"{row['Correlation']:.2f}", va='center')

        plt.show()

    # Plot GDP distribution histograms
    for religion in subset_means['Dominant'].unique():
        sub_means = subset_means[subset_means['Dominant'] == religion]

        plt.figure(figsize=(10, 5))
        sns.histplot(
            sub_means['Country_Mean_GDP'],
            bins=10,
            kde=True,
            stat='count',
            color='skyblue',
            edgecolor='black'
        )

        plt.title(f'Tier {tier} | Distribution of Country Mean GDP per Capita: {religion}')
        plt.xlabel('Country Mean GDP per Capita')
        plt.ylabel('Count of Countries')
        plt.grid(axis='y', linestyle='--', alpha=0.7)
        plt.show()
```

# **Countries by Tier Change**

The goal of this section is to determine whether or not the changes in tiers of the religions were significantly different across time. This section may show if there were points in time with significant changes and which religions had a surge. To determine this, the data must first pass a normality test with the Shapiro-WIlk test as well as equal variances through the Levene's test. Passing both leads to the use of ANOVA but if either fails, the Kruskal-Wallis test will be performed instead.

**Process**

1. Create a dataframe with the differences between the tiers across the years of each religion through .diff.
2. Test normality with Shapiro-Wilk by checking each dominant religion and their trend rows.
3. Perform Levene's test on the differences across the years.
4. If both passed, perform ANOVA but since both failed, Kruskal-Wallis's Test was used.

```python
tiers_trends = tiers_by_year.dropna()

trends = []
for religion in religion_columns:
    values = tiers_trends.query(f'Dominant == "{religion}"')['Tier_Value'].tolist()
    diffs = np.diff(values)
    trends.append(diffs)

print("\nShapiro Wilk's Normality Test")
shapiro_pvals = []
for i in range(len(religion_columns)):
    label = religion_columns[i]
    group = trends[i]
    pval = shapiro(group).pvalue
    shapiro_pvals.append(pval)
    print(f"{label}: p-value = {pval:.4}")

levene_stat, levene_p = levene(*trends)
print("\nLevene's Equal Variance Test")
print(f"Levene’s test p-value = {levene_p:.4f}")

kruskal_stat, kruskal_p = kruskal(*trends)

print(f"Kruskal-Wallis statistic = {kruskal_stat:.2f}")
print(f"p-value = {kruskal_p:.4}")

print("\nInterpretation:")
if kruskal_p <= 0.05:
    print("Reject H0 = There are significant differences in tier trends between religions.")
else:
    print("Do not reject H0 = No significant differences in tier trends between religions.")
```

Based on this analysis, the GDP tier changes for each dominant religion group were found to be normally distributed, but the Levene's test indicated unequal variances between groups. Therefore, the non-parametric Kruskal-Wallis test was used instead of ANOVA. The Kruskal-Wallis test resulted in a p-value of 0.1233, which is greater than the significance level of 0.05. This result indicates that there are no statistically significant differences in GDP tier mobility trends between the countries grouped by dominant religion.

# **Created Tier Values**
Similar to the process applied in the GDP dataset, the countries are binned for each 5 year interval through each 20% bin quantile. Their values are assigned from 1 to 5 with 5 being the highest. These values are then tracked and a mobility value is given for if they increase or decrease by rank in the next 5 years.

This will be used to find if certain religions are related to a higher growth rate as well as to see if certain religions are more represented in certain HDI tiers.



```python
clean_df = clean_df.sort_values(by=['Country', 'Year']).reset_index(drop=True)

year_range = sorted(clean_df['Year'].unique())
processed = []

for year in year_range:
    curr_group = clean_df[clean_df['Year'] == year].copy()
    hdi = curr_group['HDI']

    if hdi.nunique() < 5:
        curr_group['HDI_Tier_Value'] = 0
    else:
        try:
            curr_group['HDI_Tier_Value'] = pd.qcut(
                hdi.rank(method='first'),
                q=5,
                labels=[1, 2, 3, 4, 5]
            ).astype(int)
        except Exception as e:
            print(f"Year {year} failed: {e}")
            curr_group['HDI_Tier_Value'] = 0

    processed.append(curr_group)

tiered_hdi = pd.concat(processed, ignore_index=True)

mobility = []
prev_tiers = {}

for year in year_range:
    curr_group = tiered_hdi[tiered_hdi['Year'] == year].copy()
    curr_group = curr_group[curr_group['HDI_Tier_Value'] != 0]

    mobility_values = []

    for index, row in curr_group.iterrows():
        country = row['Country']
        current_tier = row['HDI_Tier_Value']
        previous_tier = prev_tiers.get(country)

        if previous_tier is None:
            mobility_values.append(0)
        else:
            mobility_values.append(current_tier - previous_tier)

        prev_tiers[country] = current_tier

    curr_group['HDI_Mobility_Value'] = mobility_values
    mobility.append(curr_group)

mobility_hdi = pd.concat(mobility, ignore_index=True)

hdi_progression = tiered_hdi.merge(
    mobility_hdi[['Country', 'Year', 'HDI_Mobility_Value']],
    on=['Country', 'Year'],
    how='left'
)

if 'HDI_Mobility_Value' not in hdi_progression.columns:
    hdi_progression['HDI_Mobility_Value'] = 0

clean_df = clean_df.merge(
    hdi_progression[['Country', 'Year', 'HDI_Tier_Value', 'HDI_Mobility_Value']],
    on=['Country', 'Year'],
    how='left'
)

clean_df['HDI_Tier_Value'] = clean_df['HDI_Tier_Value'].fillna(0).astype(int)
clean_df['HDI_Mobility_Value'] = clean_df['HDI_Mobility_Value'].fillna(0)

clean_df = clean_df.sort_values(by=['Country', 'Year']).reset_index(drop=True)

display(clean_df)
```

# HDI Chi-Square Test of Independence

To examine whether there is a significant association between a country's dominant religion and its level of human development, a Chi-Square Test of Independence was conducted. By categorizing countries into HDI tiers and analyzing them with their dominant religions, the test evaluates whether certain religious groups are disproportionately represented in higher or lower tiers.

**Process**

To ensure the Chi-Square test included all countries consistently across time, the group created an interpolated version of the original dataset. This was necessary because many countries had missing entries for specific years. Without interpolation, countries with fewer data points would be underrepresented.

The group filled in the missing years (e.g., 1995, 1990, etc.) by applying the following rules:
* Country: Directly reused from existing entries
* Year: Interpolated every 5 years backward until 1960
* Tier Value: Taken as the most frequent (mode) value per country
* Dominant Religion: Also selected using the mode
* Total Population: Estimated as 95% of the population from 5 years later
* Dominant Religion Population: Also decayed by 5% from the later year
* Dominant Population Share: Computed from interpolated values
* Homogeneity: Averaged across existing entries

```python
# 1. Prepare the base: get unique Country-Year-Dominant-Total Pop-Dominant Share
country_dominant = (
    religion_full_df
    .query('Year >= 1990')  # Adjust if needed
    [['Country', 'Year', 'Dominant', 'total_population', 'dominant_share']]
    .drop_duplicates()
    .dropna()
)

# Compute the actual dominant religion population
country_dominant['dominant_religion_population'] = (
    country_dominant['total_population'] * country_dominant['dominant_share']
)

# 2. Prepare your HDI tier dataset
chi_df = tiered_hdi.query('Year >= 1990').drop(['HDI'], axis=1)
chi_df = chi_df.merge(country_dominant, on=['Country', 'Year'], how="left").dropna()

# Fix column names if duplicates
chi_df = chi_df.rename(columns={'Dominant_x': 'Dominant'}).drop(columns=['Dominant_y'], errors='ignore')

# 3. Interpolate missing years backward for each country
interpolated_chi_df = pd.DataFrame()
country_list = chi_df['Country'].unique()

for country in country_list:
    sub_df = chi_df[chi_df['Country'] == country].sort_values(by='Year', ascending=False).reset_index(drop=True)

    # Mode and base values
    tier_value = sub_df['HDI_Tier_Value'].mode().iloc[0]
    dominant = sub_df['Dominant'].mode().iloc[0]
    total_pop = sub_df['total_population'].iloc[0]
    dominant_pop = sub_df['dominant_religion_population'].iloc[0]
    dominant_share = dominant_pop / total_pop if total_pop != 0 else 0
    homogeneity = sub_df['Homogeneity'].mean() if 'Homogeneity' in sub_df else None

    # Years present
    year_list = set(sub_df['Year'])

    for year in range(2020, 1989, -5):
        if year not in year_list:
            new_row = {
                'Country': country,
                'Year': year,
                'HDI_Tier_Value': tier_value,
                'Dominant': dominant,
                'total_population': total_pop,
                'dominant_religion_population': dominant_pop,
                'dominant_share': dominant_share,
                'Homogeneity': homogeneity
            }
            sub_df = pd.concat([sub_df, pd.DataFrame([new_row])], ignore_index=True)

            # Decay population backward by 5% each step
            total_pop *= 0.95
            dominant_pop *= 0.95

    sub_df = sub_df.sort_values(by='Year').reset_index(drop=True)
    interpolated_chi_df = pd.concat([interpolated_chi_df, sub_df], ignore_index=True)

# 4. Summary outputs
print('[Countries per year]')
for year in range(1990, 2021, 5):
    print(f"{year}: {len(interpolated_chi_df[interpolated_chi_df['Year'] == year])}")

print('\n[Entries per dominant religion]')
for religion in interpolated_chi_df['Dominant'].unique():
    count = len(interpolated_chi_df[interpolated_chi_df['Dominant'] == religion])
    print(f"{religion}: {count}")

display(interpolated_chi_df.head())
```

# HDI Group Mean Analysis
In this section, the variance of HDI per dominant religion were analyzed to test if there is a significant difference present. In using ANOVA, three conditions must be met:
- Normal population distribution
- Same or equal variance
- Data are independent
(Source: https://online.stat.psu.edu/stat500/lesson/10/10.2/10.2.1)




**Data Setup for Normality**

Considering that the current nature of *clean_df*, *yearly_religion_groups* was made to be a dictionary of dataframes used for testing the assumptions of ANOVA as well as correlation analysis. Organizing the data to become a dictionary of different years assigned to the respective dataframes made it easier to perform operations isolated by year.

```python
# Separated by year to test religion groups for each year's normality and equal variance
yearly_religion_groups = {
    # Dictionary pair of year and the different religion dataframes
    year: {
        # Contains the countries under the specified dominant religion
        religion: sub_group
        for religion, sub_group in group.groupby('Dominant')
    }
    for year, group in clean_df.groupby('Year')
}

```

### Step 1: Test for normality using Shapiro-Wilk Test
Interpretation:
- p >  0.05 -> the HDI values in that religion group appear to be normally distributed.
- p <= 0.05 -> not normally distributed.




```python
# Perform Shapiro-Wilk Test to assess normality of the distribution for each religion's HDI
for year, year_df in clean_df.groupby('Year'):
    print(f"\n=== Year: {year} ===")
    # Actual test for the specified year's set of religion HDI
    for religion, rel_df in year_df.groupby('Dominant'):
        stat, p = shapiro(rel_df['HDI'])
        print(f"{religion}: p = {p:.4f} -> {'Normal' if p > 0.05 else 'Not normal'}")
```

### Step 2: Test for Equal Variance using Levene's Test
#### Interpretation:
- p > 0.05 -> variances are equal thus assumptions are met
- p ≤ 0.05 -> Variances are not equal

```python
# Perform Levene's Test for each year's variance of HDI by religion
for year, religions in yearly_religion_groups.items():
    # Collect HDI arrays for each religion (only if they have at least 3 entries)
    hdi_groups = [df['HDI'].values for df in religions.values() if len(df) >= 3]
    # Actual test for variance for the HDI groups per year
    stat, p = levene(*hdi_groups)
    print(f"Year {year}")
    print(f"p = {p:.4f} -> {'Equal variances' if p > 0.05 else 'Unequal variances'}")
```

### Interpretation
Given that the p-values for Shapiro-Wilk violate the test for normality while the Levene's Test passes for equal variance, the option of a parametric test such as One-way Anova is removed. Thus the use of a non-parametric test should be implemented instead.

### Step 3: Use Spearman to assess correlation between Religion and HDI
Since the Shapiro-Wilk test indicated that the data violated normality assumptions, Spearman’s rank correlation was used to assess the monotonic relationship between religious population proportions and HDI. # Group Mean Comparison for Life Expectancy and Dominant Religion
**Process**

To be able to formally conclude that the dominant religions do perform significantly differently from each other in terms of life expectancy, it was necessary to test whether the observed differences in life expectancies across religious groups were statistically significant rather than due to random variation.

The two assumption checks were performed: the Shapiro-Wilk test for normality on each group and Levene's test for equal variances for across the groups. If all religion groups passed both assumptions, a one-way ANOVA test was applied to compare mean life expectancy. If either assumption failed, the non-parametric Kruskal-Wallis test was used instead to compare medians. The test used, along with its F- or H-statistic and corresponding p-value, was recorded for each time interval. This approach allowed the group to account for the distributional characteristics of the data before selecting the appropriate test. A line graph was later produced to visualize the H-statistics across time intervals, with visual indicators highlighting intervals where statistically significant differences were seen.

```python
lifeExpectancy_df_long
```

```python
results = []

for year in sorted(lifeExpectancy_df_long['Year'].unique()):
    subset = lifeExpectancy_df_long[lifeExpectancy_df_long['Year'] == year]
    grouped = subset.groupby('Dominant')['Life Expectancy']

    group_data = []
    group_labels = []

    for name, group in grouped:
        if len(group) > 2:
            group_data.append(group)
            group_labels.append(name)

    # Get just the p-val
    shapiro_pvals = [shapiro(g)[1] for g in group_data]
    shapiro_normal = all(p > 0.05 for p in shapiro_pvals)


    levene_p = levene(*group_data).pvalue
    levene_equal_var = levene_p > 0.05

    results.append({
        'Year': year,
        'Shapiro All Normal': shapiro_normal,
        'Levene_p': round(levene_p, 4),
        'Equal Variance': levene_equal_var,
    })

normality_results_df = pd.DataFrame(results)
display(normality_results_df)
```

Across every year, the Shapiro-Wilk tests indicate that at least one religion's life-expectancy distributions are not normally distributed. In contrast, the Levene's p-values are all above 0.05, which means the assumption of equal variances among religions is satisfied in every year.

Although equal variances hold, the normality requirement is violated for every year. Therefore, the Kruskal-Wallis test will be used to determine whether life expectancy differs across dominant religions.

```python
kruskal_results = []

for year in sorted(lifeExpectancy_df_long['Year'].unique()):
    subset = lifeExpectancy_df_long[lifeExpectancy_df_long['Year'] == year]
    grouped = subset.groupby('Dominant')['Life Expectancy']

    group_data = []
    group_labels = []

    for name, group in grouped:
        if len(group) > 2:
            group_data.append(group)
            group_labels.append(name)

    if len(group_data) < 2:
        continue

    stat, p_val = kruskal(*group_data)

    kruskal_results.append({
        'Year': year,
        'test-statistic': stat,
        'p-value': p_val,
        'Significant': p_val < 0.05,
    })

kruskal_df = pd.DataFrame(kruskal_results)
display(kruskal_df)
```

The Kruskal-Wallis test results reveal that for the majority of years, there are statistically significant differences in life expectancy across the dominant religions. This suggests that religion may be a factor that causes significant differences median life expectancy over time. Except for a few years, specifically 1991, 1993, 1994, and 2010 where the p-values remained below the 0.05 significance threshold. Overall, the dominant religion shows consistent divergences in the life expectancy distributions among the countries and is worth deeper study.

This approach is more suitable for K-Means clustering, which relies on distance calculations. Standardization ensures that no single feature disproportionately affects the clustering outcome.

```python
# To cap Liechtenstein because it is an extreme outlier and take up an entire tier of its own
cap_value = final_scaled_df['Mean_GDP_per_capita'].quantile(0.99)
print(f"\n GDP per capita cap value: {cap_value:.2f}")

final_scaled_df['Mean_GDP_per_capita'] = np.where(
    final_scaled_df['Mean_GDP_per_capita'] > cap_value,
    cap_value,
    final_scaled_df['Mean_GDP_per_capita']
)

# 5 Clusters to become the tiers
n_clusters = 5

# K-Means Algorithm
kmeans = KMeans(n_clusters=n_clusters, random_state=69)
# fit_predict does the centroids and reassignments
final_scaled_df['KMeans_Cluster'] = kmeans.fit_predict(final_scaled_df[features].values)

# The final success score, GDP and HDI are ranked equally while
# Life Expectancy is treated less because it is already a subset of HDI
final_scaled_df['Success_Score'] = (
    0.4 * final_scaled_df['Mean_GDP_per_capita'] +
    0.4 * final_scaled_df['HDI'] +
    0.2 * final_scaled_df['Life Expectancy']
)

cluster_summary = (
    final_scaled_df.groupby('KMeans_Cluster', observed=True)['Success_Score']
    .mean()
    .reset_index()
    .sort_values(by='Success_Score', ascending=False)
    .reset_index(drop=True)
)

# Prints the summary of all clusters
print("\nCluster means after capping + scaling:")
display(cluster_summary)
```

**Success Value Correlation**

In order to show the correlation between the population proportion of each dominant religion and the success score, a Spearman correlation analysis was conducted. This approach allows us to examine whether differences in the size of religious populations are  associated with changes in developmental performance across countries and over time.

```python
final_scaled_df

results = []

for religion, group in final_scaled_df.groupby('Dominant'):
    subset = group[['dominant_share', 'Success_Score']].dropna()

    if len(subset) > 1:
        corr, pval = spearmanr(subset['dominant_share'], subset['Success_Score'])
        results.append({
            'Religion': religion,
            'correlation': corr,
            'p_value': pval,
        })

correlation_results_df = pd.DataFrame(results)
correlation_results_df = correlation_results_df.sort_values(by='correlation', ascending=False)

display(correlation_results_df)
```

The correlation analysis between dominant religion share and the composite Success Score yielded statistically non-significant results across all religious groups.

While unaffiliated countries showed a moderate positive correlation (ρ = 0.32), and Buddhists a moderate negative one (ρ = -0.31), the p-values were far above the 0.05 threshold, indicating that these relationships are not statistically significant. This outcome contrasts with earlier findings where raw metrics such as GDP or HDI displayed stronger associations with religious distribution.

One key reason for this discrepancy is the nature of the Success Score measurement we decided, which is a standardized composite of GDP per capita, HDI, and life expectancy. Standardization may have affected the weight of extreme outliers such as in GDP which is why it was more vivid in its raw form but muted in the combination.

Additionally, because the Success Score blends multiple development indicators, any strong correlation with a single metric may be diluted if the other components are less aligned. These findings suggest that while religion may influence individual aspects of development, its direct association with the overall measure of success is weak. Thus, the low correlations observed here may be accurate representations of the limitations of using dominant religion share alone to explain broad, multifaceted development scores.

## **Developmental Factor Impact**

**GDP and Dominant Religion**

GDP per capita emerged as the most strongly affected development indicator by a country's dominant religion. Countries with an unaffiliated majority consistently ranked highest in GDP per capita, exhibiting both high means and medians with minimal dependence on outliers. HDI patterns aligned somewhat with those of GDP but with more nuance. Unaffiliated nations again led with the highest and most stable HDI scores. **Life Expectancy and Dominant Religion**

Life expectancy was the least affected development indicator by dominant religion. All groups exhibited upward trends over time, yet unaffiliated countries maintained the highest and most consistent life expectancies. Muslim-majority countries again ranked lowest, while Christians and Buddhists occupied the middle positions. The differences, while present, were less dramatic than in GDP or HDI. The Kruskal-Wallis test identified significant differences across religions in most years, but Spearman correlations between religious dominance and life expectancy were statistically insignificant across all groups. No outliers were detected, reinforcing the reliability of the group-level trends.

**Clustering and Composite Success Score**

To evaluate overall development success, a Success Score was constructed using standardized GDP, HDI, and life expectancy values. K-Means clustering grouped countries into five success tiers. Unaffiliated countries, though few in number, were disproportionately represented in the Top and High tiers. Muslim-majority countries were concentrated in the Low and Bottom tiers, suggesting persistent underdevelopment. Christian-majority countries were widely spread but most concentrated in the Middle tier, indicating balanced, moderate performance. Buddhist-majority countries mostly clustered in the Low tier but avoided the Bottom tier.

Despite these clear descriptive patterns, Spearman correlation tests between dominant religion share and the Success Score showed no statistically significant relationships. The composite nature of the Success Score likely diluted the strong associations seen in individual metrics, especially GDP. Additionally, small sample sizes in some groups may make the outcomes unreliable. This suggests that while religion may correlate with certain outcomes, it alone is insufficient to explain overall success.
